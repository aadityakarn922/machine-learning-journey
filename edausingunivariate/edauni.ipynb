```python
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv('train.csv')
df.head()
```

CATEGORIAL PLOT
#COUNT PLOT


```python
import matplotlib.pyplot as plt

sns.countplot(data=df, x='Survived', hue='Survived', palette=['#d95f59', '#3b8ea5'], legend=False)
plt.title('Passenger Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Number of Passengers')
plt.show()
```

```python


sns.countplot(data=df, x='Sex', hue='Sex', palette=['#d95f59', '#3b8ea5'], legend=False)
plt.show()
```
