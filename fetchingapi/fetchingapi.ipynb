```python
dfs = []
columns = ['id', 'title', 'popularity', 'release_date', 'vote_average']

for i in range(1, 6):
    response = requests.get(
        'https://api.themoviedb.org/3/movie/top_rated?api_key=41b57d5c5cbd2a75191e46ea75b72680&page={}'.format(i),
        timeout=30,
    )
    response.raise_for_status()
    temp_df = pd.DataFrame(response.json().get('results', []), columns=columns)
    dfs.append(temp_df)

df = pd.concat(dfs, ignore_index=True)
```
